```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("sales_data.csv")

# df.head()
df.tail()
# df.dtypes
# display(df.info)
metrics = ['Unit_Cost', 'Unit_Price', 'Profit', 'Cost', 'Revenue']

def orderQ(order):
    count = 0
    for item in order:
        if (item != 0 and item != ""): 
            count += 1 
    mean = round(sum(list(order))/count, 2)  
    minimym = min(order)
    maximym = max(order)
    standardDeviation = order.std()
    median = order.median() 

    #plt.figure(figsize=(8, 4))
    plt.hist(order,figsize=(8, 4))
    plt.grid(True, alpha=0.3)
    plt.show()

    
    print(f"количество элементов: {count}")
    print(f"среднее значение: {mean}")
    print(f"минимум: {minimym}")
    print(f"максимум: {maximym}")
    print(f"стандартное отклонение: {standardDeviation}")
    print(f"медиана: {median}")

    


orderQ(df['Order_Quantity'])
```
