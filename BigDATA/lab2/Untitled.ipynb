```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("sales_data.csv")

display(pd.concat([df.head(), df.tail()]))

# df.dtypes
display(df.info())
display(df.describe())

print(['Customer_Age', 'Order_Quantity', 'Unit_Cost', 'Unit_Price','Profit', 'Cost', 'Revenue'])


def orderQ(order):
    count = 0
    countOne = 0
    countTWO = 0
    for item in order:
        if (item != 0 and item != ""): 
            count += 1 
            
    mean = round(sum(list(order))/count, 2)  
    minimym = min(order)
    maximym = max(order)
    standardDeviation = order.std()
    median = order.median() 
    
    print(f"количество элементов: {count}")
    print(f"среднее значение: {mean}")
    print(f"минимум: {minimym}")
    print(f"максимум: {maximym}")
    print(f"стандартное отклонение: {standardDeviation}")
    print(f"медиана: {median}")
    


    
    plt.figure(figsize=(10, 4))
    plt.hist(order,edgecolor='black',bins = "auto",alpha=0.7)
    plt.grid(True, alpha=0.3)
    plt.yticks([0,3000, 4000, 5000, 6000, 7000, 8000, 15000, 20000])
    plt.xticks(range(1, 33, 1))
    plt.show()


    category = {}
    for item in df['Product_Category']:
        if item in category:
            category[item] += 1
        else:
            category[item] = 1
    print(category)
    plt.pie(category.values(),labels = category.keys(),autopct='%1.1f%%',startangle=90)


    



    locations = df[['Country','State']].drop_duplicates()
    locArr = np.array(locations)
    locObj = {}
    for item in locArr:
        if item[0] in locObj:
            if item[1] in locObj[item[0]]:
                continue
            else:
                locObj[item[0]].append(item[1])
        else:
            locObj[item[0]] = [item[1]]

    for key, value in locObj.items():
        print(f"{key}: {value}\n")
    

    products = sorted(list(df['Product'].drop_duplicates()))

    # for item in products:
    #     print(f"# {item}")
    print(products)

    ProductOrderQuantity =  df[['Product','Order_Quantity']]
    POQarr = np.array(ProductOrderQuantity)

    print(" ")
    
    POQobj = {}
    for item in POQarr:
        if item[0] in POQobj:
            POQobj[item[0]] += item[1]
        else:
            POQobj[item[0]] = item[1]

    tenProduct = sorted(POQobj.items(), key=lambda x: x[1], reverse=True)[:10]

    products = [item[0] for item in tenProduct]
    quantities = [item[1] for item in tenProduct]

    plt.figure(figsize=(8, 6))
    plt.bar(products, quantities, edgecolor='black',alpha=0.7)
    plt.xticks(rotation=45, ha='right')
    plt.grid(True, alpha=0.3)
    
    
   
    plt.show()
    
orderQ(df['Order_Quantity'])

    




```
